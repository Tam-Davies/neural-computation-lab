```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def sigmoid_derivative(z):
    return sigmoid(z) * (1 - sigmoid(z))
```

```python
def relu(z):
    return np.maximum(0, z)
```

```python
def relu_derivative(z):
    return np.where(z > 0, 1, 0)
```

```python
# Generate a range of input values
z = np.linspace(-10, 10, 400)
```

```python
print(z)
```

```python
sigmoid_grad = sigmoid_derivative(z)
relu_grad = relu_derivative(z)
```

```python
# Plot the activation functions
plt.figure(figsize=(12, 6))

# Plot Sigmoid and its derivative
plt.subplot(1, 2, 1)
plt.plot(z, sigmoid(z), label='Sigmoid Activation', color='b')
plt.plot(z, sigmoid_grad, label="Sigmoid Derivative", color='r', linestyle='--')
plt.title('Sigmoid Activation & Gradient')
plt.xlabel('Input Value (z)')
plt.ylabel('Activation / Gradient')
plt.legend()

# Plot ReLU and its derivative
plt.subplot(1, 2, 2)
plt.plot(z, relu(z), label='ReLU Activation', color='g')
plt.plot(z, relu_grad, label="ReLU Derivative", color='r', linestyle='--')
plt.title('ReLU Activation & Gradient')
plt.xlabel('Input Value (z)')
plt.ylabel('Activation / Gradient')
plt.legend()

plt.tight_layout()
plt.show()
```

```python

def tanh(z):
    return np.tanh(z)

def tanh_derivative(z):
    return 1 - np.tanh(z) ** 2
```

```python
# Generate a range of input values
z = np.linspace(-5, 5, 100)

# Get the Relu and tanH gradient values
tanh_grad = tanh_derivative(z)
relu_grad = relu_derivative(z)


# Plot the activation functions
plt.figure(figsize=(12, 6))

# Plot Relu and its derivative
plt.subplot(1, 2, 1)
plt.plot(z, relu(z), label='ReLU Activation', color='g')
plt.plot(z, relu_grad, label="ReLU Derivative", color='r', linestyle='--')
plt.title('ReLU Activation & Gradient')
plt.xlabel('Input Value (z)')
plt.ylabel('Activation / Gradient')
plt.legend()


# Plot tanH and its derivative
plt.subplot(1, 2, 2)
plt.plot(z, tanh(z), label='tanH Activation', color='g')
plt.plot(z, tanh_grad, label="tanH Derivative", color='r', linestyle='--')
plt.title('tanH Activation & Gradient')
plt.xlabel('Input Value (z)')
plt.ylabel('Activation / Gradient')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def sigmoid_derivative(z):
    s = sigmoid(z)
    return s * (1 - s)
```

```python
z = np.linspace(-10, 10, 200)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(z, sigmoid(z), label="Sigmoid")
plt.plot(z, relu(z), label="ReLU")
plt.title("Activation Functions")
plt.xlabel("Input (z)")
plt.ylabel("Activation")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(z, sigmoid_derivative(z), label="Sigmoid derivative")
plt.plot(z, relu_derivative(z), label="ReLU derivative")
plt.title("Activation Derivatives")
plt.xlabel("Input (z)")
plt.ylabel("Derivative")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
layers = np.arange(1, 21)

sigmoid_gradients = 0.25 ** layers
relu_gradients = 1.0 ** layers

print("Layer | Sigmoid gradient | ReLU gradient")

for layer, sigmoid_g, relu_g in zip(
    layers, sigmoid_gradients, relu_gradients
):
    print(f"{layer:5} | {sigmoid_g:.10f} | {relu_g:.1f}")
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    layers,
    sigmoid_gradients,
    marker="o",
    label="Sigmoid"
)

plt.plot(
    layers,
    relu_gradients,
    marker="o",
    label="ReLU (positive inputs)"
)

plt.title("Illustrative Gradient Propagation Across Layers")
plt.xlabel("Number of Layers")
plt.ylabel("Gradient Contribution")
plt.yscale("log")
plt.legend()
plt.grid(True)
plt.show()
```
